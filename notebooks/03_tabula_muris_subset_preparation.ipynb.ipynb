```python
import scanpy as sc
from pathlib import Path

raw_data_path = Path("/Users/palaksharma/single-cell-aging-clock/data/raw/tabula_muris_senis_dev.h5ad")

print("File exists:", raw_data_path.exists())

aging_data_preview = sc.read_h5ad(raw_data_path, backed="r")

print(aging_data_preview)
print(aging_data_preview.obs.columns.tolist())
```

```python
print(aging_data_preview)
```

```python
metadata_columns = aging_data_preview.obs.columns.tolist()

print("Number of metadata columns:", len(metadata_columns))
print(metadata_columns)
```

```python
possible_age_columns = []
possible_tissue_columns = []
possible_celltype_columns = []

for column in metadata_columns:
    col_lower = column.lower()
    
    if "age" in col_lower:
        possible_age_columns.append(column)
        
    if "tissue" in col_lower or "organ" in col_lower:
        possible_tissue_columns.append(column)
        
    if "cell" in col_lower or "type" in col_lower or "annotation" in col_lower:
        possible_celltype_columns.append(column)

print("Found these options for Age:", possible_age_columns)
print("Found these options for Tissue/Organ:", possible_tissue_columns)
print("Found these options for Cell-Type:", possible_celltype_columns)
```

```python
print(aging_data_preview.obs["free_annotation"].value_counts().head(20))
```

```python

metadata_table = aging_data_preview.obs.copy()
selected_cell_ids = []
cells_per_age_group = 2000

unique_ages = metadata_table["age"].unique()
print(f" Found these age groups to process: {unique_ages}\n")

for age in unique_ages:
    cells_in_this_age = metadata_table[metadata_table["age"] == age]
    
    total_available_cells = len(cells_in_this_age)
    print(f" Age {age}: Total available cells = {total_available_cells}")
    
    if total_available_cells > cells_per_age_group:
        number_to_take = cells_per_age_group
    else:
        number_to_take = total_available_cells
        
    print(f"   -> Sampling {number_to_take} cells from this group")
    
    sampled_subset = cells_in_this_age.sample(n=number_to_take, random_state=42)
    
    cell_ids_list = sampled_subset.index.tolist()
    selected_cell_ids.extend(cell_ids_list)


print(f" Total selected cells for our model: {len(selected_cell_ids)}")
```

```python
aging_subset = aging_data_preview[selected_cell_ids, :].to_memory()

print(aging_subset)
```
